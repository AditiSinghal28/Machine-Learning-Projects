IMPORTING THE DEPENDENCIES

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

DATA COLLECTION AND PREPROCESSING

```python
# loading dataset into dataframe
credit_card_data = pd.read_csv('/content/creditcard.csv')
```

Class : 0 --> legit

This dataset is highly unbalanced

```python
# separating the data for analysis
legit = credit_card_data[credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]
```

UNDERSAMPLING

Build a sample dataset containing similar distribution of normal(legit) transactions and fraudulent transactions

Number of fraud transactions = 492

```python
# we are going to take 492 samples of legit transactions from all the legit transactions
# and create a new dataset with 492 legit and 492 frud transactions
```

```python
legit_sample = legit.sample(492) # or legit.sample(n= 492)
```

```python
legit_sample.shape
```

Concatenating two dataframes

```python
new_dataset = pd.concat([legit_sample, fraud], axis = 0) # axis = 0 means we are joining them row wise
```

```python
new_dataset.head()
```

```python
new_dataset.shape
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

```python
new_dataset.info()
```

```python
new_dataset.describe()
```

SPLITTING THE DATASET INTO FEATURES AND TARGET

```python
X = new_dataset.drop(columns = 'Class', axis = 1)
Y = new_dataset['Class']
```

```python
print(X)
```

```python
print(Y)
```

SPLITTING THE DATASET INTO TRAINING AND TEST DATA

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, stratify = Y, random_state = 2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

MODEL TRAINING

```python
model = LogisticRegression()
```

```python
model.fit(X_train, Y_train)
```

EVALUATION

```python
# accuracy score of training data
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print('Accuracy on training data : ', training_data_accuracy)
```

```python
# accuracy score of test data
X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print('Accuracy on test data : ', test_data_accuracy)
```
