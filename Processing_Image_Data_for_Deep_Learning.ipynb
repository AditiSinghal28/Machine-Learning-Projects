Convert rgb image to grayscale image using openCV

```python
import cv2
```

```python
img = cv2.imread('/content/3000.jpg?width=1200&height=1200&quality=85&auto=format&fit=crop&s=cdadb2f0061a9d1671d268efdd5a572d')
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python
type(grayscale_img)
```

```python
grayscale_img.shape
```

```python
grayscaleplot = plt.imshow(grayscale_img, cmap = 'gray')
plt.show()
```
