```python
import numpy as np
import pandas as pd
import sklearn.datasets
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
data_frame = pd.DataFrame(breast_cancer_dataset.data, columns = breast_cancer_dataset.feature_names)
```

```python
data_frame.shape
```

```python
data_frame['target'] = breast_cancer_dataset.target
```

1 --> benign

0 --> malignant

```python
data_frame.groupby('target').mean()
```

```python
X = data_frame.drop(columns = 'target', axis = 1)
Y = data_frame['target']
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, random_state = 12, test_size = 0.2)
```

```python
X_test.head()
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Standardize the data

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_std = scaler.fit_transform(X_train)
X_test_std = scaler.transform(X_test)
```

**BUILDING** **NEURAL** **NETWORK**

```python
# importing tensorflow and keras
import tensorflow as tf
tf.random.set_seed(3)
from tensorflow import keras
```

```python
# setting up the layers of nn

model = keras.Sequential([
    keras.layers.Flatten(input_shape=(30,)), # Corrected: specify input_shape for the first layer
    keras.layers.Dense(20, activation = 'relu'),
    keras.layers.Dense(2, activation = 'sigmoid') #no of neurons in o/p layer = number of classes in target variable
])
```

```python
# compiling the nn

model.compile(optimizer = 'adam',
              loss = 'sparse_categorical_crossentropy', #for numerical encoding
              metrics = ['accuracy'])
```

```python
# training the nn

history = model.fit(X_train_std, Y_train, validation_split = 0.1, epochs = 10)
```

Visualizing accuracy and loss

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('model accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')
plt.legend(['training data', 'validation data'], loc = 'lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')
plt.legend(['training data', 'validation data'], loc = 'upper right')
```

Accuracy of model on test data

```python
loss, accuracy = model.evaluate(X_test_std, Y_test)
print("Test loss : ", loss)
print("Test accuracy : ", accuracy)
```
